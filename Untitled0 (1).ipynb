## **Importing Additional Libraries**

This cell imports `numpy` for numerical operations, `seaborn` for statistical data visualization, and `matplotlib.pyplot` for creating static, interactive, and animated visualizations in Python.

```python
try:
    df_nike = pd.read_csv('nike_5year_sales_data.csv')

    df_nike['Channel_Type'] = np.where(
        df_nike['Sales_Channel'].isin(['Nike.com', 'Nike App']),
        'Digital',
        'Physical'
    )

    print("Data loaded successfully into 'df_nike'.")
    display(df_nike.head())

except FileNotFoundError:
    print("Error: 'nike_5year_sales_data.csv' not found. Please ensure the file has been uploaded.")
except KeyError as e:
    print(f"Error: expected column {e} is missing from the CSV. Check the column names.")
except Exception as e:
    print(f"An unexpected error occurred while loading the data: {e}")
```

This cell attempts to load the `nike_5year_sales_data.csv` file into a pandas DataFrame called `df_nike`. Error handling is included to catch `FileNotFoundError` and `KeyError`.

## **Total Sales by Region**

This cell calculates the total sales for each `Region` and sorts them in descending order to identify the regions with the highest sales.

```python
# Find the total sales and total units sold for each Category in a single result (use .agg())
df_nike.groupby('Category').agg({
    'Total_Sales': 'sum',
    'Units_Sold': 'sum'
}).sort_values(by='Total_Sales', ascending=False)
```

## **Total Sales and Units Sold by Category**

This cell groups the data by `Category` and calculates the sum of `Total_Sales` and `Units_Sold` for each category, then sorts them by `Total_Sales`.

```python
# Create a pivot table with Region as rows, Category as columns, and the sum of Total_Sales as values
df_nike.pivot_table(index='Region', columns='Category', values='Total_Sales', aggfunc='sum')

```

## **Sales by Region and Category (Pivot Table)**

This cell creates a pivot table to show the `Total_Sales` for each `Region` (rows) across different `Category` types (columns).

```python
combo = (df_nike.groupby(['Region', 'Category'])['Total_Sales']
                .sum()
                .sort_values(ascending=False))

print(combo.head(1))   # top combination
print(combo.head(10))  # top 10 for context
```

## **Top Region-Category Combinations by Sales**

This cell identifies the top combinations of `Region` and `Category` based on `Total_Sales`.

```python
yearly = df_nike.groupby('Year')['Total_Sales'].sum().reset_index()
yearly['YoY_Growth_%'] = yearly['Total_Sales'].pct_change() * 100
print(yearly)
```

## **Yearly Sales and Year-over-Year Growth**

This cell calculates the total sales for each `Year` and then computes the Year-over-Year (YoY) growth percentage.

```python
monthly = df_nike.groupby('Date')['Total_Sales'].sum().reset_index()
monthly['MoM_Growth_%'] = monthly['Total_Sales'].pct_change() * 100
print(monthly)
```

```python
df_nike['Channel_Type'] = np.where(
    df_nike['Sales_Channel'].isin(['Nike.com', 'Nike App']),
    'Digital',
    'Physical'
)
```

## **Categorizing Sales Channels**

This cell re-applies the logic to categorize sales channels into 'Digital' or 'Physical' based on specific `Sales_Channel` values. This ensures the `Channel_Type` column is correctly defined for further analysis.

```python
summary = df_nike.groupby('Channel_Type').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Transactions=('Total_Sales', 'count'),
    Avg_Order_Value=('Total_Sales', 'mean')
).reset_index()

print(summary)
```

## **Sales Summary by Channel Type**

This cell groups the data by `Channel_Type` to calculate the `Total_Sales`, number of `Transactions`, and `Avg_Order_Value` for each channel type.

```python
yearly_channel = df_nike.pivot_table(index='Year', columns='Channel_Type',
                                     values='Total_Sales', aggfunc='sum')
print(yearly_channel)
```

## **Yearly Sales by Channel Type (Pivot Table)**

This cell creates a pivot table to display the `Total_Sales` for each `Year` across 'Digital' and 'Physical' `Channel_Type`s.

```python
pivot = df_nike.pivot_table(index='Region', columns='Sales_Channel',
                            values='Total_Sales', aggfunc='sum')

top_channel = pivot.idxmax(axis=1)
print(top_channel)
```

## **Top Sales Channel per Region**

This cell determines which `Sales_Channel` generates the highest `Total_Sales` in each `Region`.

```python
top10 = (df_nike.groupby('Product_Name')
                .agg(Total_Sales=('Total_Sales', 'sum'),
                     Total_Units=('Units_Sold', 'sum'))
                .assign(Avg_Selling_Price=lambda d: (d['Total_Sales'] / d['Total_Units']).round(2))
                .sort_values('Total_Sales', ascending=False)
                .head(10)
                .reset_index())

print(top10)
```

## **Top 10 Products by Sales**

This cell identifies the top 10 products based on their `Total_Sales` and also calculates their `Total_Units` sold and `Avg_Selling_Price`.

```python
top3 = (df_nike.groupby(['Category', 'Product_Name'], as_index=False)['Total_Sales'].sum()
               .sort_values(['Category', 'Total_Sales'], ascending=[True, False])
               .groupby('Category')
               .head(3))

print(top3)
```

## **Visualizing Total Sales by Product Category**

Let's visualize the total sales contribution of each product category using a bar plot. This will help us quickly understand which categories are driving the most revenue.

```python
category_sales = df_nike.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=category_sales.index, y=category_sales.values, palette='viridis')
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (USD)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## **Visualizing Total Sales by Region**

Next, let's visualize the total sales across different geographic regions. This will highlight the key markets for Nike.

```python
region_sales = df_nike.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=region_sales.index, y=region_sales.values, palette='magma')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales (USD)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## **Analyzing Sales Trend Over Time**

It's important to understand how sales have changed over the years. Let's visualize the yearly sales to spot any trends or anomalies.

```python
yearly_sales = df_nike.groupby('Year')['Total_Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Total_Sales', data=yearly_sales, marker='o')
plt.title('Total Sales Over Years')
plt.xlabel('Year')
plt.ylabel('Total Sales (USD)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(yearly_sales['Year'])
plt.tight_layout()
plt.show()
```

## **Sales Distribution by Channel Type**

Let's visualize the proportion of sales coming from 'Digital' versus 'Physical' channels to understand the distribution of sales channels.

```python
channel_type_sales = df_nike.groupby('Channel_Type')['Total_Sales'].sum()

plt.figure(figsize=(8, 8))
plt.pie(channel_type_sales, labels=channel_type_sales.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Proportion of Total Sales by Channel Type')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```

## **Distribution of Unit Prices**

Understanding the distribution of `Unit_Price` can provide insights into pricing strategies and product value.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_nike['Unit_Price'], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Unit Prices')
plt.xlabel('Unit Price (USD)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## **Relationship between Unit Price and Total Sales**

A scatter plot can help visualize if there's any correlation between the `Unit_Price` of products and their `Total_Sales`.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Unit_Price', y='Total_Sales', data=df_nike, alpha=0.6, hue='Category', size='Units_Sold', sizes=(20, 400))
plt.title('Unit Price vs. Total Sales (Colored by Category, Sized by Units Sold)')
plt.xlabel('Unit Price (USD)')
plt.ylabel('Total Sales (USD)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
